```python
import pandas as pd

# read vind_og_sol.csv

df = pd.read_csv('vind_og_sol.csv', delimiter=';')

df.head()
```

```python
# sort the dataframe by status to be 'I drift'
df = df[df['Status'] == 'I drift']
```

```python
# change 'Idriftsdato' to datetime format
df['Idriftsdato'] = pd.to_datetime(df['Idriftsdato'], format='%d-%m-%Y')

#  filter 'Idriftsdato' to be before 2026
df = df[df['Idriftsdato'] < '01-31-2025']
```

```python
# sum 'InstalleretkW'
total_installed_kw = df['InstalleretkW'].sum()

print(f'Total installed capacities of wind and solar 2025: {total_installed_kw/1000} MW')
```

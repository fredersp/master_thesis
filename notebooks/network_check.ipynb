```python
import pypsa

# Load network
network = pypsa.Network('base_s_3_elec_.nc')
```

```python
# show only generators from DK
network.generators[network.generators.bus.str.startswith('DK')]
```

```python
# sum all generating units capacities in Denmark
sum_cap = network.generators.loc[network.generators.bus.str.startswith('DK'), 'p_nom_opt'].sum()
print(sum_cap)

```

```python
# Capacity of different fuel types in Denmark
fuel_cap = network.generators.loc[network.generators.bus.str.startswith('DK')].groupby('carrier')['p_nom_opt'].sum()
print(fuel_cap)
```

```python


# find percentual shares of each fuel type in Denmark
fuel_shares = (fuel_cap / fuel_cap.sum()) * 100
print(fuel_shares)
```
